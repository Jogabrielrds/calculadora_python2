```python
def somar(a, b):
    return a + b

def subtrair(a, b):
    return a - b

def multiplicar(a, b):
    return a * b

def dividir(a, b):
  if b == 0:
    print("Erro:nao tem como dividir por 0")
  else:
    return a / b
```

```python
while True:
    print("\n=== Calculadora ===")
    print("1 - Somar")
    print("2 - Subtrair")
    print("3 - Multiplicar")
    print("4 - Dividir")
    print("0 - Sair")

    opcao = input("Escolha a operação (0-4): ")

    if opcao == "0":
        print("Até logo!")
        break

    num1 = float(input("Primeiro número: "))
    num2 = float(input("Segundo número: "))

    if opcao == "1":
        print("Resultado:", somar(num1, num2))
    elif opcao == "2":
        print("Resultado:", subtrair(num1, num2))
    elif opcao == "3":
        print("Resultado:", multiplicar(num1, num2))
    elif opcao == "4":
        print("Resultado:", dividir(num1, num2))
    else:
        print("Opção inválida")
```
